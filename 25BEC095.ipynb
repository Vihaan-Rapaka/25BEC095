## Question 1: Basic Python Operations
Enter two numbers and display their sum, difference, product, quotient, and remainder.

```python
first_number = float(input("Enter the first number: "))
second_number = float(input("Enter the second number: "))

print("Sum:", first_number + second_number)
print("Difference:", first_number - second_number)
print("Product:", first_number * second_number)

if second_number == 0:
    print("Quotient and remainder are undefined when the second number is zero.")
else:
    print("Quotient:", first_number / second_number)
    print("Remainder:", first_number % second_number)
```

## Question 2: Lists and Tuples
Work with a list of ten numbers: add, remove, and insert an item, find the maximum and minimum, then convert it to a tuple.

```python
numbers = []
for number in range(10):
    value = int(input(f"Enter number {number + 1} of 10: "))
    numbers.append(value)

print("Starting list:", numbers)
numbers.append(int(input("Number to add: ")))

remove_value = int(input("Number to remove: "))
if remove_value in numbers:
    numbers.remove(remove_value)
else:
    print("That number is not in the list.")

insert_value = int(input("Number to insert: "))
position = int(input(f"Position to insert at (0 to {len(numbers)}): "))
if 0 <= position <= len(numbers):
    numbers.insert(position, insert_value)
else:
    print("That position is outside the list.")

print("Updated list:", numbers)
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("As a tuple:", tuple(numbers))
```

## Question 3: Strings
Count the characters, vowels, and consonants, reverse the string, and check whether it is a palindrome.

```python
text = input("Enter a string: ")
letters = [character.lower() for character in text if character.isalpha()]
vowels = sum(character in "aeiou" for character in letters)
consonants = len(letters) - vowels
clean_text = "".join(character.lower() for character in text if character.isalnum())

print("Characters:", len(text))
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Reversed:", text[::-1])
print("Palindrome:", clean_text == clean_text[::-1])
```

## Question 4: Sets and Dictionaries
Show common set operations, then record five students' marks and display the student with the highest mark.

```python
set_a = {int(value.strip()) for value in input("Set A (integers separated by commas): ").split(",")}
set_b = {int(value.strip()) for value in input("Set B (integers separated by commas): ").split(",")}

print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
print("Symmetric difference:", set_a ^ set_b)

student_marks = {}
for number in range(5):
    name = input(f"Student {number + 1} name: ").strip()
    student_marks[name] = float(input(f"{name}'s marks: "))

best_student = max(student_marks, key=student_marks.get)
print("Student marks:", student_marks)
print(f"Highest marks: {best_student} ({student_marks[best_student]:g})")
```

## Question 5: GPA Calculator
Calculate GPA using the weighted grade-point formula: total of grade point times credit, divided by total credits.

```python
course_count = int(input("How many courses? "))
while course_count <= 0:
    print("Enter at least one course.")
    course_count = int(input("How many courses? "))

weighted_points = 0.0
total_credits = 0.0
for number in range(course_count):
    print(f"Course {number + 1}")
    grade_point = float(input("Grade point: "))
    credits = float(input("Credits: "))
    while credits <= 0:
        print("Credits must be greater than zero.")
        credits = float(input("Credits: "))
    weighted_points += grade_point * credits
    total_credits += credits

print(f"GPA: {weighted_points / total_credits:.2f}")
```

## Question 6: Functions and Iteration
Use separate functions for factorial, the Fibonacci series, and checking whether a number is prime.

```python
def factorial(number):
    if number < 0:
        return None
    result = 1
    for value in range(2, number + 1):
        result *= value
    return result


def fibonacci(term_count):
    series = []
    first, second = 0, 1
    for _ in range(term_count):
        series.append(first)
        first, second = second, first + second
    return series


def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


number = int(input("Number for factorial: "))
answer = factorial(number)
if answer is None:
    print("Factorial is not defined for negative numbers.")
else:
    print("Factorial:", answer)

term_count = int(input("Number of Fibonacci terms: "))
if term_count < 0:
    print("The number of terms cannot be negative.")
else:
    print("Fibonacci series:", fibonacci(term_count))

prime_check = int(input("Number to check for prime: "))
print("Prime:", is_prime(prime_check))
```

## Question 8: Exception Handling
Divide two numbers and handle a zero denominator with `try` and `except`.

```python
numerator = float(input("Numerator: "))
denominator = float(input("Denominator: "))

try:
    result = numerator / denominator
except ZeroDivisionError:
    print("Division by zero is not allowed.")
else:
    print("Result:", result)
```

Create the three requested lists using list comprehensions.

## Question 10: Student Record Menu
Use a dictionary to add, update, search, delete, and display student records during the current session.

```python
students = {}

while True:
    print("\n1. Add student")
    print("2. Update marks")
    print("3. Search for student")
    print("4. Delete student")
    print("5. Display all students")
    print("0. Exit")
    choice = input("Choose an option: ").strip()

    if choice == "1":
        name = input("Student name: ").strip()
        students[name] = float(input("Marks: "))
        print("Student added.")
    elif choice == "2":
        name = input("Student name to update: ").strip()
        if name in students:
            students[name] = float(input("New marks: "))
            print("Marks updated.")
        else:
            print("Student not found.")
    elif choice == "3":
        name = input("Student name to search: ").strip()
        if name in students:
            print(f"{name}: {students[name]:g}")
        else:
            print("Student not found.")
    elif choice == "4":
        name = input("Student name to delete: ").strip()
        if name in students:
            del students[name]
            print("Student deleted.")
        else:
            print("Student not found.")
    elif choice == "5":
        if students:
            for name, marks in students.items():
                print(f"{name}: {marks:g}")
        else:
            print("No student records yet.")
    elif choice == "0":
        break
    else:
        print("Choose a number from 0 to 5.")
```
